```python
import time
import numpy as np

def encontrar_vazio(tabuleiro):
    for linha in range(9):
        for coluna in range(9):
            if tabuleiro[linha, coluna] == 0:
                return linha, coluna
    return None


def valor_valido(tabuleiro, linha, coluna, valor):
    # Linha
    if valor in tabuleiro[linha, :]:
        return False

    # Coluna
    if valor in tabuleiro[:, coluna]:
        return False

    # Bloco 3x3
    inicio_linha = (linha // 3) * 3
    inicio_coluna = (coluna // 3) * 3

    bloco = tabuleiro[
        inicio_linha:inicio_linha + 3,
        inicio_coluna:inicio_coluna + 3
    ]

    if valor in bloco:
        return False

    return True


def resolver_backtracking(tabuleiro):
    posicao = encontrar_vazio(tabuleiro)

    # Não existem mais células vazias
    if posicao is None:
        return True

    linha, coluna = posicao

    for valor in range(1, 10):

        if valor_valido(tabuleiro, linha, coluna, valor):

            tabuleiro[linha, coluna] = valor

            if resolver_backtracking(tabuleiro):
                return True

            # Retrocede
            tabuleiro[linha, coluna] = 0

    return False


def executar_backtracking(sudoku):
    tabuleiro = sudoku.copy()

    inicio = time.perf_counter()

    solucionado = resolver_backtracking(tabuleiro)

    fim = time.perf_counter()

    return {
        "solucao": tabuleiro,
        "solucionado": solucionado,
        "tempo": fim - inicio
    }
```

```python
# ==========================================================
# EXECUÇÃO DO BACKTRACKING
# ==========================================================

instancias = {
    "Fácil": sudoku_facil,
    "Médio": sudoku_medio,
    "Difícil": sudoku_dificil
}

resultados_backtracking = {}


for dificuldade, sudoku in instancias.items():

    print("\n" + "=" * 60)
    print(f"BACKTRACKING - SUDOKU {dificuldade.upper()}")
    print("=" * 60)

    resultado = executar_backtracking(sudoku)

    resultados_backtracking[dificuldade] = resultado

    print(f"Solucionado: {resultado['solucionado']}")
    print(f"Tempo: {resultado['tempo']:.6f} s")
    print(f"Tempo: {resultado['tempo'] * 1000:.3f} ms")

    print("\nSolução:")
    print(resultado["solucao"])


# ==========================================================
# RESUMO FINAL
# ==========================================================

print("\n")
print("=" * 80)
print("RESUMO FINAL DO BACKTRACKING")
print("=" * 80)

for dificuldade, resultado in resultados_backtracking.items():

    print(f"\nSudoku {dificuldade}")
    print("-" * 50)

    print(f"Solucionado: {resultado['solucionado']}")
    print(f"Tempo: {resultado['tempo']:.6f} s")
    print(f"Tempo: {resultado['tempo'] * 1000:.3f} ms")
```
